```python
import shutil
from pathlib import Path

import queryearth as qe

import vizqueryearth as vq
```

## Draw one shape

Draw a polygon or a rectangle with the tools on the left of the map. A
rectangle becomes a box. Draw across the date line if you like: queryearth
splits the shape at ±180.

```python
d = vq.draw()
d
```

```python
print(d.shape)  # run after drawing
rule = qe.Rule(
    code="MINE001",
    kind="warning",
    title="Example",
    message="A rule whose shape was drawn on the map.",
    scope=qe.Scope(space=d.space),
)
print(rule.to_feature()["geometry"])
```

## Edit the shapes of a rule file

The editor works on a copy of the demo rules. Pick a rule in the list or click
its shape, then draw, drag or delete shapes on the map. **New rule** adds a rule
with the next free code. **Save** checks every rule with queryearth and writes the
file. Drawing to the map's top or bottom edge reaches the pole, as the Arctic rule
DEMO003 does.

```python
path = Path("my_rules.geojson")  # a copy, next to this notebook
shutil.copy("demo_rules.geojson", path)
ed = vq.Editor(path)
ed
```

The same steps from code:

```python
ed.new()  # DEMO005, selected
ed.set_shape(vq.arctic(75))
ed.save()
print(qe.Registry("check").load(path))
```
